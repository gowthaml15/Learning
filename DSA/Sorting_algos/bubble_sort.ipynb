### Bubble Sort
Simple sorting algorithm that compares the adjacent elements and swaps them if they are in the wrong order. <br>
Algortim Steps:<br>
- Start at the begining of the list.
- Compare adjacent elements : if the current element is greater than the next element, swap them.
- Move to the next pair of elements and repeat the comparison and swapping process if necessary.
- Continue this process for the entire list.
- After one complete pass, the largest element will have moved to its correct position at the end.
- Repeat the process for remaining elements, ignoring the last sorted elements.
- Continue until no swaps are needed, indicating the list is sorted.
<br>
Time Complexity:<br>
- Best case (Already sorted): O(n)
- Average Case: O(n^2)
- Worst Case (Reverse order) : O(n^2)
<br>
Space Complexity:
O(1)

```python
arr = [64, 34, 25, 12, 22, 11, 90]
len(arr)
```

```python
for i in range(len(arr)-1):
    for j in range(len(arr)-1-i):
        if arr[j]>arr[j+1]:
            arr[j+1],arr[j] = arr[j], arr[j+1]
print(arr)
```
